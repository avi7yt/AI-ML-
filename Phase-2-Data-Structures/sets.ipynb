```python
# 1. Creating a set with elements
fruits = {"apple", "banana", "cherry"}

# Duplicates are automatically ignored
numbers = {1,2,2,3,4,4}
print(numbers)

# CRITICAL: To create an EMPTY set, you must use set().
# Using {} create an empty dictionary!
empty_set = set()
empty_dict = {}
```

```python
# 2. Modifying a Set

colors = {"red", "green"}

colors.add("blue")
colors.update(["yellow", "orange"])
colors.remove("red")
colors.discard("purple")

print(colors)
```

```python
# 3. Core Mathematical Set Operations
set_a = {1,2,3,4}
set_b = {3,4,5,6}

print(set_a.union(set_b))
print(set_a.intersection(set_b))
print(set_a.difference(set_b))
print(set_a.symmetric_difference(set_b))
```

```python
# 4. Common Use Cases

# Membership Testing
guest = {"Alice", "Bob", "Charlie"}
print("Alice" in guest)

# Removing Duplicates From a List
raw_items = [1,2,2,3,4,4,4,5]
unique_items = list(set(raw_items))
print(unique_items)
```
