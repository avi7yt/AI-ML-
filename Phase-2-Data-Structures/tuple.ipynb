```python
# Core syntac and creation 
#Standard tuple
user_data = ("Alice", 28, "Engineer")

#CRITICAL: A single-item tuple REQUIRES a trailing comma
single_item = ("Apple",)

print(user_data)
print(single_item)
```

```python
# Common Operations
# 1. Accessing Items (Indexing & Slicing)
fruits = ("Apple", "Banana", "Cherry", "Orange")
print(fruits[0])
print(fruits[-1])
print(fruits[1:2])

# 2. Tuple Unpacking
coordinates = (4,10)
x,y = coordinates 
print(x,y)

# Using an asterisk (*) to gather remaining items
items = (1,2,3,4,5)
first, *middle, last = items
print(first)
print(middle)
print(last)

# 3. Built-in Methods
print(items.count(3)) #return number of items a value appears
print(items.index(4)) # return index
```
