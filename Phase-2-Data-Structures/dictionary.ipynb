```python
# Core Syntax and Concepts  

#  1. Creating a Dictionary
user = {"name": "Alice", "age": 25, "role": "Admin"}
user2 = dict(name = "Alice", age= 25, role="Admin")

print(user)
print(user2)

# 2. Accessing Items
print(user["name"])
print(user.get("email", "Not Found"))

# 3. Adding and Updating Items
user["email"] = "alice@gmail.com"
user["age"] = 26
print(user)

# 4. Removing Items
del user["role"]
age = user.pop("age")
user.popitem()

print(user)

# 5. Essential Dictionary Methods
user = {"mob": "630605XXXX", "Delivery Address": "51B Lotus Shrishti, Crossing Republic"}

print(user.keys())
print(user.values())
print(user.items())
print(user.update(user2))
user.clear()
print(user)
```

```python
# Iterating Through a Dictionary
user = {"name": "Alice", "age":25, "email": "alice@gmail.com"}
for key,value in user.items():
    print(f"{key}: {value}")

for key in user.keys():
    print(key)

# Dictionary Comprehensions
# x = 1
square = {x: x**2 for x in range(1,5)}
print(square)
```
