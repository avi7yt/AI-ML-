```python
# Essential List Methods
books = [
        "To Kill a Mockingbird by Harper Lee",
        "1984 by George Orwell",
        "Pride and Prejudice by Jane Austen",
        "The Great Gatsby by F.ScottFitzgerald"
        ]

# Addint Items: 
books.append("Moby-Dick by Herman Melville")
books.insert(2,"Jane Eyre by Charlotte Bronte")
books.extend([
    "Crime and Punishment by Fyodor Dostoevsky",
    "War and Peace by Leo Tolstoy",
    "The Odyssey by Homer",
    "The Catcher in the Tye by J.D. Salinger",
    "The Hobbit by J.R.R. Tolkien"])

#Removing Items: 
books.pop()
books.remove("The Catcher in the Tye by J.D. Salinger")
# books.clear()

print(books)

#Utility Operations:
print(len(books))
books.sort()
print(books)

#List Slices
booksSub = books[1:4]
print(booksSub)
```
