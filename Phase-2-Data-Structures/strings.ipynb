```python
# 1. Creating and Printing Strings

# single line string
text1 = 'Hello'
text2 = "World"

# Multiline string
multiline = """This is a
multiline string."""

print(text1)
print(text2)
print(multiline)

# 2. Indexing and Slicing

word = "python"
#Indexing

print(word[0])
print(word[-1])

#Slicing syntax -> string[start:end:step]
print(word[0:2])
print(word[2:])
print(word[::-1])

# Concatenation and Repetition
greeting = "Hello" + " " + "World"
repeat = "Hi!" * 3
print(greeting)
print(repeat)
```

```python
# String Methods

# 1. Changing Text Case
text1 = "python"
print(text1.upper())
text2 = "PyThOn"
print(text2.lower())
text3 = "hello world"
print(text3.capitalize())
print(text3.title())

text4 = "PyThOn"
print(text4.swapcase())

# 2. Searching and Replacing
print("banana".count("an"))
print("coding".find("d")) #return index otherwise -1
print("coding".index("i")) #it throws an error if not exist
print("I love Java".replace("Java", "Python"))

# 3. Splitting, Joining, and Trimming
print("a,b,c".split(","))
print("-".join(['2026','08','06']))
print("   clean me    ".strip())
print("   clean me    ".lstrip())
print("   clean me    ".rstrip())

# 4. Content Validation (The is Methods)
text5 = "Hello everyone"
text6 = text5.split(" ")
print("".join(text6).isalpha())

text7 = "12334455"
print(text7.isdigit())

text8 = "2222"
print(text8.isalnum())

text9 = "     "
print(text9.isspace())
print(text8.isspace())

text10 = "avivishwakarma7yt@gmail.com"
print(text10.startswith("avi"))
print(text10.endswith("@gmail.com"))
```
