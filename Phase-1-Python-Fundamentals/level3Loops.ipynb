```python
# 12. Print
# using for loop
for i in range(1,101): print(i)

# uring while loop
count = 1
while count <= 100:
    print(count)
    count += 1  #updates condition
```

```python
# 13. Print Reverse
for i in range(100, 0, -1) : print(i)
```

```python
# 14. Sum of First N Numbers
num = int(input("Enter a number: "))
sum = 0
for i in range(num+1):sum += i

print(f"Sum of first {num} numbers is {sum}")
```

```python
# 15. Multiplication Table
tableNumber = int(input("Enter number to print table: "))
for i in range(1,11): print(f"{tableNumber} x {i} = {tableNumber*i}")
```

```python
# 16. Factorial
factorialNumber = int(input("Enter number to print factorial: "))
fact = 1
for i in range(1,factorialNumber+1): fact*=i
print(fact)
```
