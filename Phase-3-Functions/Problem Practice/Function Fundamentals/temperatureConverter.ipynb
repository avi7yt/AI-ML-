```python
# C to K: K = C+273.15
# K to C: C = K-273.15
# F to C: C = (F-32)*5/9
# C to F: F = C(9/5)+32
# F to K: K = (F-32)*5/9+273.15
# K to F: F = (K - 273)*9/5+32

def celsius_to_fahrenheit(temp):
    return temp*(9/5)+32

def fahrenheit_to_celsius(temp):
    return (temp-32)*(5/9)

def celsius_to_kelvin(temp):
    return temp + 273.15

def kelvin_to_celsius(temp):
    return temp - 273.15

def temperature_conversion():
    temp = float(input("Enter temperature: "))
    unit = input("Enter unit celsius-C, faherenheit-F, kelvin-K: ")

    match unit:
        case 'C':
            print(f"Fahrenheit: {celsius_to_fahrenheit(temp)}")
            print(f"Kelvin: {celsius_to_kelvin(temp)}")
        case 'F':
            print(f"Celsius: {fahrenheit_to_celsius(temp)}")
            print(f"Kelvin: {celsius_to_kelvin(fahrenheit_to_celsius(temp))}")
        case 'K':
            print(f"Celsius: {kelvin_to_celsius(temp)}")
            print(f"Fahrentheit: {celsius_to_fahrenheit(kelvin_to_celsius(temp))}")

temperature_conversion()
```
