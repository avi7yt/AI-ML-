```python
# Basic Function
def greet():
    print("Hello, Akash")
greet()
```

```python
# Function with Parameters
def greet(name):
    print("Hello",name)

greet("Akash")
greet("Rohit")

# Multiple Parameters
def add(a, b):
    print(a+b)
    
add(10,20)

# Default Argument
def greetUser(name = "User"):
    print(f"Hello, {name}")

greetUser()

# Keyword Argument
def student(name, age):
    print(name, age)

student(age=20, name="Akash")

# Positional Arguments
#  - normally, arguments according to their position match hota hai

# *args
def total(*args):
    sum = 0
    for a in args:
        sum += a
    
    print(sum)

total(10,20,30,40)

# **kwargs
def library(**book):
    myBook = book
    print(myBook)

library(book1 = "Being a Human", book2 = "Udan", book3 = "Rangoli", book4 = "Primary Goal")

# Combining parameters
def combiningParameters(a, b = 10, *arr, **emp):
    salary = a * b
    print(salary)
    print(arr)
    print(emp)

combiningParameters(10, 3,4,5, Manager= "Rohit", SeniorDeveloper = "Amish")

# Return Statement
def subtract(a,b):
    return a - b

subtract(10,3)

# Lambda Function
square = lambda x: x*x
print(square(4))

# Recursive Function
def rec(count):
    while count > 0:
        print(count)
        count -= 1
        rec(count)

rec(10)

# Function as a Variable
def sayHello(user):
    print("Hello", user)

hi = sayHello("Vikram")

# Type-Hints
def typeHints(a:int, b:int)->int:
    return a + b
print(typeHints(22,22))
```
